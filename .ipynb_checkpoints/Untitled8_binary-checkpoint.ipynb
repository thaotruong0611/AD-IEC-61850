```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
import numpy as np
from sklearn.preprocessing import StandardScaler, OrdinalEncoder, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
```

```python
data = pd.read_csv("Data.csv")
```

```python
data["dif_stNum"] = data["dif_stNum"].astype("int64")
data["dif_sqNum"] = data["dif_sqNum"].astype("int64")
```

```python
x = data.drop(columns = ["Label"])
y = data[["Label"]]
```

```python
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
y_numeric = label_encoder.fit_transform(y)
print(label_encoder.classes_)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
   x, y_numeric, test_size = 0.3, random_state = 42 ,shuffle=True, stratify = y_numeric)
```

```python
num_transformer = Pipeline(steps=[
    # ("imputer", SimpleImputer(strategy="median")),
    ("num_scaler", StandardScaler())
])
```

```python
nom_transformer = Pipeline(steps=[
    # ("imputer", SimpleImputer(strategy="most_frequent")),
    # ("HashingEncoder", HashingEncoder(n_components=2))
    ("encoder", OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])
```

```python
# # stNum = X_train["stNum"].unique()
# Dif_stNum = X_train["Dif_stNum"].unique()
# Dif_sqNum = X_train["Dif_sqNum"].unique()
Circuit_breaker_Open_Close_Status_level = [np.sort(X_train['Circuit breaker Open/Close Status'].unique()).tolist()]
```

```python
ord_transformer = Pipeline(steps=[
    # ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OrdinalEncoder(categories = Circuit_breaker_Open_Close_Status_level, handle_unknown='use_encoded_value', unknown_value=-1))
])
```

```python
preprocessor = ColumnTransformer([
    ("num_feature", num_transformer, ['Length', 'heartbeat', 'I-high VT', 'I-high VT- Tran', 'I-low VT- Tran', 'I-amongst-feeders', "sqNum"]),
    ("ord_feature", ord_transformer, ['Circuit breaker Open/Close Status']),
    ("nom_feature", nom_transformer, ['stNum', 'Source', 'dif_stNum', 'dif_sqNum','goID'])
])
```

```python
from sklearn.svm import SVC

from sklearn.model_selection import GridSearchCV
```

```python
type(X_train)
```

```python
reg = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("regressor", SVC() )
])
```

```python
params = {
    "regressor__C":[10, 60, 100],
    "regressor__kernel":["linear", "poly", "rbf", "sigmoid", "precomputed"],
    "regressor__gamma":["scale", "auto"],}

model = GridSearchCV(
    estimator=reg,
    param_grid=params,
    cv=3
)
```

```python
model.fit(X_train, y_train)
```

```python
type(X_train)
```

```python
np.unique(y)
```

```python
model.best_params_
```

```python
# lấy mô hình tốt nhất
best_model = model.best_estimator_
```

```python
y_predict = best_model.predict(X_test)
```
