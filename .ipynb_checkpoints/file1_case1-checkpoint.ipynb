```python
import pandas as pd
import math, random
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
import numpy as np
from sklearn.preprocessing import StandardScaler, OrdinalEncoder, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression

import matplotlib as mpl
mpl.rcParams['axes.formatter.useoffset'] = False
mpl.rcParams['axes.formatter.use_locale'] = False
mpl.rcParams['axes.formatter.use_mathtext'] = False
# --------------------------
# 0) Setup
# --------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ---- thêm các dòng này ngay sau import ----
torch.manual_seed(0)       # cố định seed cho PyTorch
np.random.seed(42)         # cố định seed cho NumPy
random.seed(0)             # cố định seed cho Python random
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False
# -------------------------------------------

from collections import Counter
```

```python
data = pd.read_csv("Data.csv")
Counter(data["Label"])
```

```python
# 1) Tạo subset anomaly nhưng KHÔNG reset index
subset_list = []

for cls in data["Label"].unique():
    if cls == "Normal":
        continue

    n_samples = 1 if cls == "Data Integrity" else 2
    take_n = min(n_samples, (data["Label"] == cls).sum())

    # ⛔️ KHÔNG reset_index ở đây
    subset = data[data["Label"] == cls].sample(n=take_n, random_state=42)
    subset_list.append(subset)

small_df = pd.concat(subset_list, axis=0)



```

```python
small_df
```

```python
small_df1 = small_df
```

```python
small_df1
```

```python
# Lưu 9 anomaly
small_df1["Label"] = (small_df1["Label"].astype(str).str.strip() != "Normal").astype(int)

small_df1.to_csv("9_anormal.csv", index=False)
print("9 anomaly", Counter(small_df1["Label"]))

# 2) Lấy phần còn lại — drop bằng index GỐC
remaining_df = data.drop(small_df.index)

print("Phan con lai", Counter(remaining_df["Label"]))
```

```python
Counter(remaining_df["Label"])
```

```python
# --- 3) Từ phần còn lại, lấy 800 Normal ---
normal_800 = remaining_df[remaining_df["Label"] == "Normal"].sample(
    n=800, random_state=42
)

# Lưu file 800 
normal_800["Label"] = (normal_800["Label"].str.strip() != "Normal").astype(int)
normal_800.to_csv("800_normal.csv", index=False)

print("\nĐã lưu 800 Normal vào 800_normal.csv")
print(normal_800.shape)
```

```python

# --- 2) Lấy phần còn lại (REMOVE 9 anomaly samples vừa chọn) ---
remaining_df1 = remaining_df.drop(normal_800.index)
```

```python
Counter(remaining_df["Label"])
```

```python

# --- 3) Từ phần còn lại, lấy 800 Normal ---
Dos_9990 = remaining_df[remaining_df["Label"] == "DoS"].sample(
    n=9990, random_state=42
)

# Lưu file Dos_9990
Dos_9990["Label"] = (Dos_9990["Label"].str.strip() != "Normal").astype(int)
Dos_9990.to_csv("Dos_9990.csv", index=False)

print("\nĐã lưu 9990 DoS vào Dos_9990.csv")
print(Dos_9990.shape)
```

```python
remaining_df2 = remaining_df1.drop(Dos_9990.index)
```

```python
Counter(remaining_df2["Label"])
```

```python
# df_normal: chỉ lấy Normal
df_normal = remaining_df2[remaining_df2["Label"] == "Normal"].copy()

# df_anomaly: phần còn lại (không phải Normal)
df_anomaly = remaining_df2[remaining_df2["Label"] != "Normal"].copy()

print("Kích thước df_normal:", df_normal.shape)
print("Kích thước df_anomaly:", df_anomaly.shape)
print("\nCount df_anomaly:")
print(df_anomaly["Label"].value_counts())
```

```python
# 1) Lấy 150 mẫu anomaly làm val/test
df_anomaly_val_test = df_anomaly.sample(n=148, random_state=42)

# 2) Phần còn lại là anomaly unlabeled
df_anomaly_unlabel = df_anomaly.drop(df_anomaly_val_test.index)

# 1) Lấy đúng 2 sample Data Integrity từ df_anomaly_unlabel
two_DI = df_anomaly_unlabel[df_anomaly_unlabel["Label"] == "Data Integrity"].sample(
    n=2, random_state=42
)

# 2) Thêm 2 sample này vào df_anomaly_val_test
df_anomaly_val_test = pd.concat([df_anomaly_val_test, two_DI], axis=0)

# 3) Xóa 2 sample đó khỏi df_anomaly_unlabel
df_anomaly_unlabel = df_anomaly_unlabel.drop(two_DI.index)

# 4) Reset index (không bắt buộc)
df_anomaly_val_test = df_anomaly_val_test.reset_index(drop=True)
df_anomaly_unlabel = df_anomaly_unlabel.reset_index(drop=True)

# Kiểm tra
print("Val/Test anomaly:", df_anomaly_val_test.shape)
print("Unlabeled anomaly:", df_anomaly_unlabel.shape)

print("\nVal/Test labels:")
print(df_anomaly_val_test["Label"].value_counts())

print("\nUnlabeled labels:")
print(df_anomaly_unlabel["Label"].value_counts())
```

```python
# df_normal hiện đang chứa 31,000 Normal
print(df_normal.shape)

# 1) Lấy 2100 mẫu làm unlabeled
df_normal_unlabeled = df_normal.sample(n=21000, random_state=42)

# 2) Phần còn lại là labeled
df_normal_labeled = df_normal.drop(df_normal_unlabeled.index)

# Kiểm tra
print("Unlabeled:", df_normal_unlabeled.shape)
print("Labeled:", df_normal_labeled.shape)
```

```python
unlabeled_pool = pd.concat([df_normal_unlabeled, Dos_9990, df_anomaly_unlabel], axis = 0)
```

```python
unlabeled_pool["Label"] = (unlabeled_pool["Label"].str.strip() != "Normal").astype(int)
```

```python
unlabeled_pool.to_csv("unlabeled_pool.csv", index=False)
```

```python
labeled_val_test_pool = pd.concat([df_normal_labeled, df_anomaly_val_test], axis = 0)
```

```python
print(labeled_val_test_pool["Label"].value_counts())
```

```python
from sklearn.model_selection import train_test_split

# labeled_val_test_pool: DataFrame đã chứa cả normal + anomaly có nhãn
print(labeled_val_test_pool["Label"].value_counts())

# 40% validation, 60% test, stratify theo Label
val_df, test_df = train_test_split(
    labeled_val_test_pool,
    test_size=0.6,                        # 60% cho test
    stratify=labeled_val_test_pool["Label"],
    random_state=42
)

print("Validation size:", val_df.shape)
print("Test size:", test_df.shape)

print("\nLabel counts in VAL:")
print(val_df["Label"].value_counts())

print("\nLabel counts in TEST:")
print(test_df["Label"].value_counts())
```

```python
val_df["Label"] = (val_df["Label"].str.strip() != "Normal").astype(int)
test_df["Label"] = (test_df["Label"].str.strip() != "Normal").astype(int)
```

```python

val_df.to_csv("val_labeled_40.csv", index=False)

test_df.to_csv("test_labeled_60.csv", index=False)
```
