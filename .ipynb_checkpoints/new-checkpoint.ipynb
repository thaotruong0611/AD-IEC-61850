```python
# # ===== Mixed-type aware: lấy chỉ số numeric & block one-hot từ ColumnTransformer =====
# preproc = reg.named_steps["preprocessor"]
# ohe     = preproc.named_transformers_["nom1_feature"].named_steps["encoder"]

# numeric_cols = ["Length", "Heartbeat"]
# num_dim      = len(numeric_cols)        # = 2

# # Số chiều của từng biến danh mục theo đúng thứ tự ["Source","Dif_stNum","Dif_sqNum","stNum"]
# cat_sizes = [len(c) for c in ohe.categories_]  # ví dụ [N1, N2, N3, N4]

# # Tạo danh sách lát cắt (start, end) cho mỗi block one-hot trong vector đã transform
# cat_blocks = []
# start = num_dim
# for sz in cat_sizes:
#     cat_blocks.append((start, start+sz))
#     start += sz

# numeric_idx = list(range(num_dim))  # [0, 1]

# # Đảm bảo dtype cho toàn bộ tensor dùng cdist
# Xn_tr_full = Xn_tr_full.float()
# Xa_tr_full = Xa_tr_full.float()

# # ===== Hàm sinh X' đúng kiểu mixed =====
# def _fill_onehot_blocks(xp, xn, xa, cat_blocks, mode="pick", w=None):
#     """Ghi các block one-hot vào xp. mode='pick' (khuyên dùng) hoặc 'argmax'."""
#     B = xn.size(0)
#     for (s, e) in cat_blocks:
#         if mode == "pick":
#             choose_n = (torch.rand(B, 1, device=xn.device) < 0.5).expand(B, e - s)
#             xp[:, s:e] = torch.where(choose_n, xn[:, s:e], xa[:, s:e])
#         else:
#             # trộn rồi argmax để trả về one-hot hợp lệ
#             mix = w * xn[:, s:e] + (1.0 - w) * xa[:, s:e]
#             idx = torch.argmax(mix, dim=1)
#             xp[:, s:e] = 0.0
#             xp[torch.arange(B, device=xn.device), s + idx] = 1.0
#     return xp

# def generate_intermediate_mixed(
#     x_n, x_a, Xn_all, Xa_all,
#     numeric_idx, cat_blocks,
#     w_low=0.3, w_high=0.7,      # w ngẫu nhiên cho từng mẫu
#     k1=50, k2=10,
#     mode="pick"                 # 'pick' (khuyến nghị) hoặc 'argmax'
# ):
#     """
#     Sinh X' theo x' = w*x_n + (1-w)*x_a cho các cột số;
#     với one-hot: chọn nguyên block từ x_n hoặc x_a (pick) hoặc trộn rồi argmax.
#     Sau đó lọc theo Algorithm 1 bằng KNN trên toàn tập.
#     """
#     B, D = x_n.size()
#     # ghép cặp 1-1
#     idx = torch.randperm(x_a.size(0), device=x_a.device)
#     xa  = x_a[idx]

#     # w riêng cho từng mẫu
#     w = torch.rand(B, 1, device=x_n.device) * (w_high - w_low) + w_low

#     # khởi tạo xp và trộn các cột số
#     xp = torch.zeros_like(x_n)
#     if len(numeric_idx) > 0:
#         ni = torch.tensor(numeric_idx, device=x_n.device, dtype=torch.long)
#         xp[:, ni] = w * x_n[:, ni] + (1.0 - w) * xa[:, ni]

#     # điền các block one-hot
#     xp = _fill_onehot_blocks(xp, x_n, xa, cat_blocks, mode=mode, w=w)

#     # --- Lọc theo Algorithm 1 ---
#     d_self_n = (xp - x_n).norm(dim=1)
#     d_self_a = (xp - xa).norm(dim=1)

#     Dn = torch.cdist(xp, Xn_all)  # [B, Nn]
#     Da = torch.cdist(xp, Xa_all)  # [B, Na]

#     k1 = min(k1, Dn.size(1))
#     k2 = min(k2, Da.size(1))
#     d_k1, _ = torch.kthvalue(Dn, k1, dim=1)
#     d_k2, _ = torch.kthvalue(Da, k2, dim=1)

#     mask = (d_k1 >= d_self_n) & (d_k2 >= d_self_a)
#     return xp[mask]

```

```python
# --------------------------
# 2) Model: encoder / decoder
# --------------------------
latent_dim = 4
in_dim = Xn_tr_full.shape[1]
out_dim=in_dim
hid = 256  
class Encoder(nn.Module):
    def __init__(self, in_dim=in_dim, hid=hid, z=latent_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_dim, hid*2), nn.ReLU(),
            nn.Linear(hid*2, hid), nn.ReLU(),
            nn.Linear(hid, z),
        )
    def forward(self, x): return self.net(x)

class Decoder(nn.Module):
    def __init__(self, out_dim=out_dim, hid=hid, z=latent_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(z, hid), nn.ReLU(),
            nn.Linear(hid, hid*2), nn.ReLU(),
            nn.Linear(hid*2, out_dim),
        )
    def forward(self, z): return self.net(z)

f = Encoder().to(device)
g = Decoder().to(device)

#cat daon tren don the dem xuongc
# --------------------------
# 3) Truncated Gaussians Dz, Dz0, Dzt
# --------------------------
def sample_trunc_gauss(n, d, sigma, clip_mult=3.0, device="cpu"):
    z = torch.randn(n, d, device=device) * sigma
    norms = z.norm(dim=1, keepdim=True)
    max_r = clip_mult * sigma
    scale = torch.clamp(max_r / (norms + 1e-8), max=1.0)
    return z * scale

sigma_n = 0.1   # Dz (normal)   — như paper
sigma_m = 0.5   # Dz0 (mid)
sigma_a = 1.0   # Dzt (abnormal)


# --------------------------
# 4) MMD^2 with Gaussian kernel (median heuristic)
# --------------------------
def _median_gamma(x, y):
    # base gamma = 1/(2*median^2); dùng multi-scale x0.5, x1, x2
    xy = torch.cat([x.detach(), y.detach()], dim=0)
    if xy.size(0) < 2:
        return (1.0,)
    # pdist trả về khoảng cách L2, không bình phương
    med = torch.pdist(xy).median()
    med = med.clamp_min(1e-6)
    base_gamma = 1.0 / (2.0 * (med**2))
    return (0.5*base_gamma, 1.0*base_gamma, 2.0*base_gamma)

def gaussian_kernel(x, y, gammas=None):
    xx = x.unsqueeze(1)  # [n,1,d]
    yy = y.unsqueeze(0)  # [1,m,d]
    d2 = ((xx - yy)**2).sum(-1)  # [n,m]
    if gammas is None:
        gammas = _median_gamma(x, y)
    K = 0
    for g in gammas:
        K = K + torch.exp(-g * d2)
    return K

# def mmd2(x, y, gammas=None):
#     n, m = x.size(0), y.size(0)
#     Kxx = gaussian_kernel(x, x, gammas)
#     Kyy = gaussian_kernel(y, y, gammas)
#     Kxy = gaussian_kernel(x, y, gammas)
#     mmd = (Kxx.sum() - Kxx.diag().sum())/(n*(n-1)+1e-12) \
#         + (Kyy.sum() - Kyy.diag().sum())/(m*(m-1)+1e-12) \
#         - 2*Kxy.mean()
#     return mmd
def mmd2(x, y, gammas=None):
    # Tính gammas 1 lần từ (x,y) rồi dùng cho cả 3 kernel
    if gammas is None:
        gammas = _median_gamma(x, y)  # ví dụ trả về tuple (g1,g2,g3)

    n, m = x.size(0), y.size(0)
    Kxx = gaussian_kernel(x, x, gammas)
    Kyy = gaussian_kernel(y, y, gammas)
    Kxy = gaussian_kernel(x, y, gammas)

    # Unbiased estimator; fallback khi n hoặc m = 1
    if n > 1 and m > 1:
        mmd = (Kxx.sum() - Kxx.diag().sum())/(n*(n-1)) \
            + (Kyy.sum() - Kyy.diag().sum())/(m*(m-1)) \
            - 2*Kxy.mean()
    else:
        mmd = Kxx.mean() + Kyy.mean() - 2*Kxy.mean()  # biased

    return mmd
# --------------------------
# 5) Intermediate generation + filtering (Algorithm 1, KNN trên toàn tập)
# --------------------------
K1_TRAIN, K2_TRAIN, W = 50, 1, 0.5
def generate_intermediate_full(x_n, x_a, Xn_all, Xa_all, w=W, k1=K1_TRAIN, k2=K2_TRAIN):
    """
    x_n: [B,d] batch normals
    x_a: [B,d] batch anomalies (đã ghép cặp ngẫu nhiên)
    Xn_all, Xa_all: toàn bộ tập train để tính KNN (ổn định hơn trong-batch)
    Trả về: x' đã vượt qua điều kiện lọc Algorithm 1
    """
    # random permutation để ghép cặp 1-1
    idx = torch.randperm(x_a.size(0), device=x_a.device)
    xa = x_a[idx]
    xp = w * x_n + (1 - w) * xa  # convex mix

    # khoảng cách tới "chính" hai điểm cấu thành x' (không cần cdist)
    d_self_n = (xp - x_n).norm(dim=1)
    d_self_a = (xp - xa).norm(dim=1)

    # KNN tới toàn bộ normal/anomaly train
    Dn = torch.cdist(xp, Xn_all)  # [B, Nn]
    Da = torch.cdist(xp, Xa_all)  # [B, Na]

    k1 = min(k1, Dn.size(1))
    k2 = min(k2, Da.size(1))
    d_k1, _ = torch.kthvalue(Dn, k1, dim=1)
    d_k2, _ = torch.kthvalue(Da, k2, dim=1)

    mask = (d_k1 >= d_self_n) & (d_k2 >= d_self_a)
    return xp[mask]



# MMD(x,x) ~ 0
x = torch.randn(256, 8)
print(mmd2(x, x).item())

# MMD tăng khi khác phân phối
y = torch.randn(256, 8) * 2.0
print(mmd2(x, y).item())


# --------------------------
# 6) Training
# --------------------------
opt = optim.Adam(list(f.parameters()) + list(g.parameters()), lr=1e-3)
mse = nn.MSELoss()

lambda1 = 2.0   # weight anomalies -> Dzt
lambda2 = 1.0   # weight intermediates -> Dz0
eta = 0.5      # reconstruction on normals

epochs = 5
f.train(); g.train()

Xa_tr_tensor = Xa_tr_full  # alias on device

for ep in range(1, epochs+1):
    total = 0.0; steps = 0; used_mid = 0
    for (xb,) in ld_loader_n:
        # print("xb", xb)
        xb = xb.to(device).float()
        # sample anomalies same size (with replacement)
        idx = torch.randint(0, Xa_tr_tensor.size(0), (xb.size(0),), device=device)
        xa = Xa_tr_tensor[idx].float()

        # generate & filter intermediates (full-dataset KNN)
        xmid = generate_intermediate_full(xb, xa, Xn_tr_full, Xa_tr_full, w=W, k1=K1_TRAIN, k2=K2_TRAIN)

        # encode
        zn = f(xb)
        za = f(xa)
        zmid = f(xmid) if xmid.numel() > 0 else None

        # target draws
        M = max(zn.size(0), 512)
        Z_n = sample_trunc_gauss(512, latent_dim, sigma_n, device=device)
        Z_a = sample_trunc_gauss(512, latent_dim, sigma_a, device=device)
        if zmid is not None:
            Z_m = sample_trunc_gauss(zmid.size(0), latent_dim, sigma_m, device=device)

        # losses
        loss_n = mmd2(zn, Z_n)
        loss_a = mmd2(za, Z_a) * lambda1
        loss_mid = mmd2(zmid, Z_m) * lambda2 if zmid is not None else torch.tensor(0.0, device=device)
        recon = mse(g(zn), xb) * eta

        loss = loss_n + loss_a + loss_mid + recon
        opt.zero_grad(); loss.backward(); opt.step()

        total += float(loss.detach()); steps += 1
        used_mid += (0 if zmid is None else zmid.size(0))

    if ep % 5 == 0 or ep == 1:
        avg_mid = used_mid / max(steps,1)
        print(f"[{ep:02d}/{epochs}] loss={total/steps:.4f} | avg_intermediate_used={avg_mid:.1f}")



import math, torch, numpy as np
with torch.no_grad():
    zn = f(Xn_all.to(device).float()); za = f(Xa_all.to(device).float())

def stats(name, Z, sigma):
    r = Z.norm(dim=1).cpu()
    print(f"{name:10s} mean||z||={r.mean():.3f}  (~{sigma*math.sqrt(2):.3f})  "
          f"p90={r.quantile(0.9):.3f}")
stats("Dz (norm)", zn, 0.1)
stats("Dzt(anom)", za, 1.0)

if zn.size(1)==2:
    c = np.corrcoef(zn[:,0].cpu(), zn[:,1].cpu())[0,1]
    print("corr(z1,z2) =", float(c))











```

```python
# --------------------------
# 7) Scoring & threshold (percentile trên normal)
#    s(x) = ||f(x)|| / sigma_n  (chuẩn hoá theo σ của Dz)
# --------------------------
f.eval()
with torch.no_grad():
    s_norm = (f(Xn_tr.to(device)).norm(dim=1) / sigma_n).cpu()
    thresh = 20  # ví dụ 90th percentile
    Xn_te = torch.from_numpy(Xn_te).float()
    Xa_te = torch.from_numpy(Xa_te).float()
    Xtest = torch.cat([Xn_te, Xa_te]).to(device)
    s_test = (f(Xtest).norm(dim=1) / sigma_n).cpu()

    y_test = torch.cat([torch.zeros(len(Xn_te)), torch.ones(len(Xa_te))])
    y_pred = (s_test > thresh).float()

# quick precision/recall/F1
tp = ((y_pred==1) & (y_test==1)).sum().item()
fp = ((y_pred==1) & (y_test==0)).sum().item()
fn = ((y_pred==0) & (y_test==1)).sum().item()
prec = tp / (tp+fp+1e-8); rec = tp / (tp+fn+1e-8); f1 = 2*prec*rec/(prec+rec+1e-8)
print(f"Threshold={thresh:.3f} | Precision={prec:.3f} Recall={rec:.3f} F1={f1:.3f}")
# ====== end ======
```

```python

# ==== VISUALIZE: Normal / Anomaly / Intermediate ====
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

@torch.no_grad()
def collect_intermediates_full(Xn_tr_full, Xa_tr_full,
                               tries=60, B=512, w=W, k1=K1_TRAIN, k2=K2_TRAIN, target_min=150):
    """
    Gom các điểm trung gian (Intermediate) qua nhiều mini-batch
    để đủ số lượng hiển thị. Tăng k1/k2 giúp bớt chặt, ra nhiều điểm hơn.
    """
    xs = []
    for _ in range(tries):
        idx_n = torch.randint(0, Xn_tr_full.size(0), (B,), device=device)
        idx_a = torch.randint(0, Xa_tr_full.size(0), (B,), device=device)
        xb, xa = Xn_tr_full[idx_n], Xa_tr_full[idx_a]
        xm = generate_intermediate_full(xb, xa, Xn_tr_full, Xa_tr_full, w=w, k1=k1, k2=k2)
        if xm.numel() > 0:
            xs.append(xm.detach().cpu())
        if sum(x.size(0) for x in xs) >= target_min:
            break
    return torch.cat(xs, dim=0) if xs else torch.empty(0, 2)

# Thu thập một lượng điểm Intermediate vừa đủ để nhìn rõ trên hình
X_mid = collect_intermediates_full(Xn_tr_full, Xa_tr_full, tries=60, B=512,
                                   w=W, k1=K1_TRAIN, k2=K2_TRAIN, target_min=150)




```
