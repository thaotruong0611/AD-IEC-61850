```python
# ============================================
# ONE-CLASS SVM cho dữ liệu IEC 61850 (Data.csv)
# - TRAIN/VAL/TEST đọc từ: label_pool / val_labeled_40 / test_labeled_60
# - TRAIN: giữ toàn bộ NORMAL + chỉ 4 ANOMALY (phần còn lại sang TEST)
# - OCSVM: train CHỈ TRÊN 800 NORMAL từ TRAIN
# - Tuning (nu, percentile q) theo F1_macro trên VALIDATION
# - Đánh giá trên TEST: AUROC (score liên tục) + F1_macro/Confusion (theo threshold tau*)
# ============================================
from sklearn.metrics import (
    roc_auc_score, f1_score, classification_report,
    confusion_matrix, ConfusionMatrixDisplay, accuracy_score
)
import math, random
from collections import Counter

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OrdinalEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.svm import OneClassSVM

import matplotlib.pyplot as plt

# ------------------------------------------------
# 0. FIX SEED
# ------------------------------------------------
np.random.seed(42)
random.seed(0)

# ------------------------------------------------
# 1. ĐỌC DATA TỪ CÁC CSV ĐÃ CHIA SẴN
# ------------------------------------------------
train_df = pd.read_csv("label_pool.csv")
val_df   = pd.read_csv("val_labeled_40.csv")
test_df  = pd.read_csv("test_labeled_60.csv")

# Split features/labels
FEATURES = [c for c in train_df.columns if c != "Label"]
X_train, y_train = train_df[FEATURES], train_df["Label"]
X_val,   y_val   = val_df[FEATURES],   val_df["Label"]
X_test,  y_test  = test_df[FEATURES],  test_df["Label"]

print("\n🔹 Sau split lần 1:")
print("   TRAIN:", X_train.shape, Counter(y_train))
print("   VAL  :", X_val.shape,   Counter(y_val))
print("   TEST :", X_test.shape,  Counter(y_test))

# ------------------------------------------------
# 4. (Giữ nguyên TRAIN/VAL/TEST như đã đọc từ CSV)
#    (Ở phiên bản này không còn di chuyển anomaly dư/DoS như bản Data.csv gốc)
# ------------------------------------------------
rng = np.random.default_rng(42)

Xtr0 = X_train.copy()
ytr0 = y_train.copy()

idx_a_tr = ytr0.index[ytr0 == 1]
idx_n_tr = ytr0.index[ytr0 == 0]

# Chỉ giữ tối đa 4 anomaly trong TRAIN (nếu bạn vẫn muốn enforce 4 anomaly)
n_keep = min(4, len(idx_a_tr))
keep_a = pd.Index(rng.choice(idx_a_tr, size=n_keep, replace=False))
drop_a = idx_a_tr.difference(keep_a)

# (Ở đây mình KHÔNG chuyển anomaly dư sang test nữa vì bạn đang dùng split cố định.
#  Chỉ đơn giản là giữ nguyên train_df như đã load. Nếu bạn muốn đúng logic
#  “4 anomaly trong train, phần dư sang test” thì có thể UNCOMMENT đoạn dưới.)
#
# X_train = pd.concat([Xtr0.loc[idx_n_tr], Xtr0.loc[keep_a]], axis=0)
# y_train = pd.concat([ytr0.loc[idx_n_tr], ytr0.loc[keep_a]], axis=0)
#
# X_test = pd.concat([X_test, Xtr0.loc[drop_a]], axis=0)
# y_test = pd.concat([y_test, ytr0.loc[drop_a]], axis=0)

# Reset index
X_train = X_train.reset_index(drop=True)
y_train = y_train.reset_index(drop=True)
X_test  = X_test.reset_index(drop=True)
y_test  = y_test.reset_index(drop=True)

print("\n🔹 SAU khi chỉnh TRAIN/TEST (nếu có):")
print("   TRAIN:", X_train.shape, Counter(y_train))
print("   VAL  :", X_val.shape,   Counter(y_val))
print("   TEST :", X_test.shape,  Counter(y_test))

# ------------------------------------------------
# 5. TIỀN XỬ LÝ (ColumnTransformer)
# ------------------------------------------------
num_features = [
    "Length", "heartbeat", "I-high VT", "I-high VT- Tran",
    "I-low VT- Tran", "I-amongst-feeders", "sqNum"
]

Circuit_breaker_Open_Close_Status_level = [
    np.sort(X_train["Circuit breaker Open/Close Status"].unique()).tolist()
]

nom_features = ["stNum", "Source", "dif_stNum", "dif_sqNum", "goID"]

num_transformer = Pipeline(steps=[
    ("num_scaler", StandardScaler())
])

ord_transformer = Pipeline(steps=[
    ("encoder", OrdinalEncoder(
        categories=Circuit_breaker_Open_Close_Status_level,
        handle_unknown="use_encoded_value",
        unknown_value=-1
    ))
])

nom_transformer = Pipeline(steps=[
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num_feature", num_transformer, num_features),
        ("ord_feature", ord_transformer, ["Circuit breaker Open/Close Status"]),
        ("nom_feature", nom_transformer, nom_features),
    ],
    remainder="drop"
)

reg = Pipeline(steps=[("preprocessor", preprocessor)])
reg.fit(X_train)

X_train_proc = reg.transform(X_train)
X_val_proc   = reg.transform(X_val)
X_test_proc  = reg.transform(X_test)

print("\n🔧 Sau preprocess:")
print("   TRAIN:", X_train_proc.shape)
print("   VAL  :", X_val_proc.shape)
print("   TEST :", X_test_proc.shape)

# ------------------------------------------------
# 6. ONE-CLASS SVM: TRAIN CHỈ TRÊN 800 NORMAL
# ------------------------------------------------
idx_n_train = np.flatnonzero(y_train.values == 0)

# lấy tối đa 800 normal đầu tiên
n_use = min(800, len(idx_n_train))
X_train_normal = X_train_proc[idx_n_train[:n_use]]

print("\n🟢 Số NORMAL dùng train OCSVM:", X_train_normal.shape[0])
print("🔴 Số ANOMALY có trong TRAIN (không dùng cho OCSVM):",
      np.sum(y_train.values == 1))

# ------------------------------------------------
# 7. TUNING (nu, percentile q) THEO F1_macro TRÊN VALIDATION
# ------------------------------------------------
candidate_nu = [0.2]

coarse = np.linspace(0.80, 0.98, 10, endpoint=True)
fine   = np.linspace(0.98, 0.999, 15, endpoint=True)
q_grid = np.unique(np.concatenate([coarse, fine]))

best_f1_macro = -1.0
best_nu = None
best_q = None
best_tau = None
best_model = None

y_val_np = y_val.values.astype(int)

print("\n🔍 Tuning (nu, percentile q) theo F1_macro trên VALIDATION:")

for nu in candidate_nu:
    # 1) Train OCSVM với nu
    oc = OneClassSVM(kernel="rbf", gamma="scale", nu=nu)
    oc.fit(X_train_normal)

    # 2) Score anomaly trên VALIDATION
    scores_val_raw = oc.decision_function(X_val_proc)  # lớn hơn = normal
    scores_val = -scores_val_raw                        # lớn hơn = anomaly

    for q in q_grid:
        tau = np.percentile(scores_val, q * 100.0)

        # anomaly = score >= tau
        y_pred_val = (scores_val >= tau).astype(int)

        # >>> F1_macro trên validation <<<
        f1_macro = f1_score(
            y_val_np,
            y_pred_val,
            average="macro",
            zero_division=0
        )

        print(f"  nu={nu:.3f}, q={q:.4f} → F1_val_macro={f1_macro:.4f}")

        if f1_macro > best_f1_macro:
            best_f1_macro = f1_macro
            best_nu = nu
            best_q = q
            best_tau = tau
            best_model = oc

print("\n✅ Best trên VALIDATION (theo F1_macro):")
print(f"   nu*        = {best_nu}")
print(f"   q*         = {best_q:.4f}")
print(f"   tau*       = {best_tau:.6f}")
print(f"   F1_val*    = {best_f1_macro:.4f} (macro)")

# ------------------------------------------------
# 8. ĐÁNH GIÁ TRÊN TEST (DÙNG (nu*, tau*) TỪ VAL)
# ------------------------------------------------
oc_best = best_model

# 1) Score anomaly trên TEST
scores_test_raw = oc_best.decision_function(X_test_proc)
scores_test = -scores_test_raw  # lớn hơn = anomaly

y_test_np = y_test.values.astype(int)

# 2) AUROC (không phụ thuộc threshold)
auroc_test = roc_auc_score(y_test_np, scores_test)

# 3) Phân loại TEST bằng threshold tau* đã chọn từ VALIDATION
y_pred_test = (scores_test >= best_tau).astype(int)

print("\n📈 KẾT QUẢ TRÊN TEST (OCSVM + threshold percentile):")
print("-----------------------------------------------------")
print(f"nu*          = {best_nu}")
print(f"q*           = {best_q:.4f}")
print(f"tau*         = {best_tau:.6f}")
print(f"AUROC_test   = {auroc_test * 100:.3f}%\n")

# 4) F1-score (macro) & Accuracy trên TEST
f1_macro_test = f1_score(
    y_test_np,
    y_pred_test,
    average="macro",
    zero_division=0
)
acc_test = accuracy_score(y_test_np, y_pred_test)

# 5) Confusion matrix trên TEST
cm = confusion_matrix(y_test_np, y_pred_test)
print("\n🧩 Confusion matrix trên TEST (rows = true, cols = pred):")
print(cm)

# ======= VẼ CONFUSION MATRIX =======
fig, ax = plt.subplots(figsize=(6, 6))

disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["Normal", "Anomaly"])
disp.plot(cmap="Blues", colorbar=True, ax=ax)

plt.title("Confusion Matrix (Test)")
plt.xlabel("Predicted label")
plt.ylabel("True label")
plt.tight_layout()
plt.show()

print("\n📈 TÓM TẮT KẾT QUẢ TRÊN TEST (chọn theo VAL F1_macro):")
print("-----------------------------------------------------")
print(f"nu*              = {best_nu}")
print(f"q*               = {best_q:.4f}")
print(f"tau*             = {best_tau:.6f}")
print(f"AUROC_test       = {auroc_test * 100:.3f}%")
print(f"F1_test (macro)  = {f1_macro_test:.4f}")
print(f"ACC_test         = {acc_test:.4f}\n")
```
