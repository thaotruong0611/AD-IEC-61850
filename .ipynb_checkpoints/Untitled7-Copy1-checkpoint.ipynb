```python
def add_physic_to_net(net_feas, index,  CB_main_path, npi_high_VT_path, i_low_VT_tran_path, i_high_VT_tran_path, i_feed1, i_feed2, i_feed3):
    cb_main = pd.read_csv(CB_main_path)
    if isinstance(index, list):# cho truong hop normal
        cb_main = cb_main.iloc[index].reset_index(drop=True)
        df10 = pd.read_csv(i_high_VT_path)
        df10 = df10.iloc[index].reset_index(drop=True)
        df10["I-high VT"] = df10[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]].mean(axis = 1)
        # print("df10", df10["I-high VT"])
        
        df30 = pd.read_csv(i_low_VT_tran_path)
        df30 = df30.iloc[index].reset_index(drop=True)
        df30["I-low VT- Tran"] = df30[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]].mean(axis=1)
        # print("df30", df30["I-low VT- Tran"])
        
        df13 = pd.read_csv(i_high_VT_tran_path)
        df13 =df13.iloc[index].reset_index(drop=True)
        
        # Tạo cột trung bình của 3 cột Current
        df13["I-high VT- Tran"] = df13[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]].mean(axis = 1)
        
        
        df31 = pd.read_csv(i_feed1)
        df31 = df31.iloc[index].reset_index(drop=True)
        df32 = pd.read_csv(i_feed2)
        df32 = df32.iloc[index].reset_index(drop=True)
        df33 = pd.read_csv(i_feed3)
        df33 = df33.iloc[index].reset_index(drop=True)
        df31["I-amongst-feeders"] = (df31[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]] + df32[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]] + df33[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]]).mean(axis=1)
        burst_df = net_feas
        if goose_frame_path.endswith("UFIED_trace.csv"):
            cb_main['Circuit breaker Open/Close Status'] = 1
        merged_df = pd.concat([burst_df, cb_main['Circuit breaker Open/Close Status'], df10["I-high VT"], df30["I-low VT- Tran"], df13["I-high VT- Tran"], df31["I-amongst-feeders"]], axis=1)

        df = merged_df.drop(columns=['integer', 'No.', 'index', 'group_time', 'Time'])
        df['Label'] = 'Normal'

        

    
    else: # cho th disturbance
        # cb_main = cb_main.iloc[[index]]
        # df10 = pd.read_csv(i_high_VT_path)
        # df10 = df10.iloc[[index]]
        # df10["I-high VT"] = df10[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]].mean(axis = 1)
        # # print("df10", df10["I-high VT"])
        
        # df30 = pd.read_csv(i_low_VT_tran_path)
        # df30 = df30.iloc[[index]]
        # df30["I-low VT- Tran"] = df30[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]].mean(axis=1)
        # # print("df30", df30["I-low VT- Tran"])
        
        # df13 = pd.read_csv(i_high_VT_tran_path)
        
        # df13 =df13.iloc[[index]]
        # # Tạo cột trung bình của 3 cột Current
        # df13["I-high VT- Tran"] = df13[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]].mean(axis = 1)
        
        
        # df31 = pd.read_csv(i_feed1)
        # df31 = df31.iloc[[index]]
        # df32 = pd.read_csv(i_feed2)
        # df32 = df32.iloc[[index]]
        # df33 = pd.read_csv(i_feed3)
        # df33 = df33.iloc[[index]]
        # df31["I-amongst-feeders"] = (df31[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]] + df32[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]] + df33[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]]).mean(axis=1)
        # burst_df = net_feas
        # if goose_frame_path.endswith("UFIED_trace.csv"):
        #     cb_main['Circuit breaker Open/Close Status'] = 1
        # merged_df = pd.concat([burst_df, cb_main['Circuit breaker Open/Close Status'], df10["I-high VT"], df30["I-low VT- Tran"], df13["I-high VT- Tran"], df31["I-amongst-feeders"]], axis=1)
        # # print(cb_main['Circuit breaker Open/Close Status'])
        # df = merged_df.drop(columns=['integer', 'No.', 'index', 'group_time', 'Time'])
        # df['Label'] = 'Fault'

        cb_main = cb_main.iloc[[index]]
        df10 = pd.read_csv(i_high_VT_path)
        df10 = df10.iloc[[index]]
        df10["I-high VT"] = df10[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]].mean(axis = 1)
        # print("df10", df10["I-high VT"])
        
        df30 = pd.read_csv(i_low_VT_tran_path)
        df30 = df30.iloc[[index]]
        df30["I-low VT- Tran"] = df30[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]].mean(axis=1)
        # print("df30", df30["I-low VT- Tran"])
        
        df13 = pd.read_csv(i_high_VT_tran_path)
        
        df13 =df13.iloc[[index]]
        # Tạo cột trung bình của 3 cột Current
        df13["I-high VT- Tran"] = df13[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]].mean(axis = 1)
        
        
        df31 = pd.read_csv(i_feed1)
        df31 = df31.iloc[[index]]
        df32 = pd.read_csv(i_feed2)
        df32 = df32.iloc[[index]]
        df33 = pd.read_csv(i_feed3)
        df33 = df33.iloc[[index]]
        df31["I-amongst-feeders"] = (df31[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]] + df32[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]] + df33[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]]).mean(axis=1)
        burst_df = net_feas
        if goose_frame_path.endswith("UFIED_trace.csv"):
            cb_main['Circuit breaker Open/Close Status'] = 1
        merged_df = pd.concat([burst_df.reset_index(drop=True), cb_main['Circuit breaker Open/Close Status'].repeat(len(burst_df)).reset_index(drop=True), df10["I-high VT"].repeat(len(burst_df)).reset_index(drop=True), df30["I-low VT- Tran"].repeat(len(burst_df)).reset_index(drop=True), df13["I-high VT- Tran"].repeat(len(burst_df)).reset_index(drop=True), df31["I-amongst-feeders"].repeat(len(burst_df)).reset_index(drop=True)], axis=1)
        # print(cb_main['Circuit breaker Open/Close Status'])
        df = merged_df.drop(columns=['integer', 'integer.1', 'integer.2', 'No.', 'group_time', 'Time'])
        df['Label'] = 'Fault'

        
    return df
```

```python
def to_csv(df, CB_main_path):
    new_path = CB_main_path.replace(".csv", "_A.csv")
    print(new_path)
    df.to_csv(new_path, index=False)
```

```python
# List all .csv in folder and all subfolders inside it
def path_pathtrace(path):
    path_files = []
    for root, dirs, files in os.walk(path):
        # print(root)
        # print(dirs)
        # print(files)
        for file in files:
            if file.endswith(".csv") and not file.endswith(("trace.csv", "_A.csv")):
                path_file = os.path.join(root, file)
                # Get the label from the top-level folder
                # relative_path = os.path.relpath(path_file, directory)
                # top_level_folder = relative_path.split(os.sep)[0]
    
                path_files.append(path_file)
                # labels_D8.append(top_level_folder)
    # Thay thế ".csv" bằng "trace.csv"
    updated_path_files = [file.replace(".csv", "_trace.csv") for file in path_files]
    
    # In kết quả
    path_files_trace =[]
    for path in updated_path_files:
        path_files_trace.append(path)
    return path_files, path_files_trace
```

```python
def dm(path_files, path_files_trace, path_all_trace, No_list, adding):
    
    cb_main = pd.read_csv(path_files)
    df = pd.read_csv(path_files_trace)
    
    # Giả sử bạn có 2 DataFrame: df (chứa dữ liệu đầy đủ) và cb_main (chứa các giá trị để tìm)
    # df có các cột: ["sqNum", "stNum", "goID", ...]
    # cb_main cũng có các cột tương tự (hoặc ít nhất 3 cột "sqNum", "stNum", "goID"]).
    
    # Tạo danh sách để lưu các hàng tìm được
    rows_to_concat = []
    
    for i in range(len(No_list)):
        value_to_find1 = df["sqNum"].iloc[i]
        value_to_find2 = df["stNum"].iloc[i]
        value_to_find3 = df["goID"].iloc[i]
        
        # Lọc ra index của những hàng trong df khớp với 3 giá trị cần tìm
        matching_indices = cb_main.index[
            (cb_main["sqNum"] == value_to_find1) &
            (cb_main["stNum"] == value_to_find2) &
            (cb_main["goID"] == value_to_find3)
        ].tolist()
        
        # Nếu tìm được index, lấy các hàng đó
        if len(matching_indices) > 0:
            matched_rows = cb_main.loc[matching_indices]
            # Thêm DataFrame con vào danh sách
            rows_to_concat.append(matched_rows)
        
        # Nối tất cả các hàng tìm được thành 1 DataFrame duy nhất
        if rows_to_concat:
            df_concatenated = pd.concat(rows_to_concat, ignore_index=True)
        else:
            print("Không tìm thấy hàng nào khớp.")

    
    
    # df_concatenated["I-high VT"] = df[["integer", "integer.1", "integer.2"]].mean(axis = 1)
    print("df.columns.tolist()", df.columns.tolist())
    df_concatenated["sqNum"] = df["sqNum"]
    df_concatenated["goID"] = df["goID"]
    df_concatenated["stNum"] = df["stNum"]
    df_concatenated["Source"] = df["Source"]
    df_concatenated["Length"] = df["Length"]
    print("df10", df10["I-high VT"])
    
    df_concatenated["Label"] = "Data Integrity"
    df_concatenated
    
    
    
    df_all = pd.read_csv(path_all_trace)
    df_all['Heartbeat'] = df_all.groupby('goID')['Time'].diff()
    m = df_all[df_all["No."].isin(No_list)]
    df_concatenated["Heartbeat"] = m["Heartbeat"].values  # duovc
    
    new_path = path_files_trace.replace("_trace.csv", adding)
    df_concatenated.to_csv(new_path, index=False)
    return df_concatenated
```

```python
def ms1_1(path_files, path_files_trace, path_all_trace, No_list, adding):
    cb_main = pd.read_csv(path_files)
    df = pd.read_csv(path_files_trace)
    df_all = pd.read_csv(path_all_trace)
    df_all['Heartbeat'] = df_all.groupby('goID')['Time'].diff()
    df_all['Dif_sqNum'] = df_all.groupby('goID')['sqNum'].diff()
    df_all['Dif_stNum'] = df_all.groupby('goID')['stNum'].diff()
    print("cb_main", cb_main)
    print("df_all", df_all)
    print("df",df )
    indices = df_all.index[df_all["No."].isin(No_list)].tolist()
    print("indeces", indices)
    indices2 = [i - 1 for i in indices]
    
    # Giả sử bạn có 2 DataFrame: df (chứa dữ liệu đầy đủ) và cb_main (chứa các giá trị để tìm)
    # df có các cột: ["sqNum", "stNum", "goID", ...]
    # cb_main cũng có các cột tương tự (hoặc ít nhất 3 cột "sqNum", "stNum", "goID"]).
    
    # Tạo danh sách để lưu các hàng tìm được
    rows_to_concat = []
    result_list = []
    for i in indices2:
        print("indexxxx", i)
        value_to_find1 = df_all["sqNum"].iloc[i]
        value_to_find2 = df_all["stNum"].iloc[i]
        value_to_find3 = df_all["goID"].iloc[i]
        
        # Lọc ra index của những hàng trong df khớp với 3 giá trị cần tìm
        matching_indices = cb_main.index[
            (cb_main["sqNum"] == value_to_find1) &
            (cb_main["stNum"] == value_to_find2) &
            (cb_main["goID"] == value_to_find3)
        ].tolist()

        print("matching_indices", len(matching_indices))
        # Nếu tìm được index, lấy các hàng đó
        if len(matching_indices) > 0:
            df_all.loc[i+1, 'Circuit breaker Open/Close Status'] = cb_main.loc[matching_indices, 'Circuit breaker Open/Close Status'].values
            df_all.loc[i+1, 'I-high VT'] = cb_main.loc[matching_indices, 'I-high VT'].values
            df_all.loc[i+1, 'I-high VT'] = df_all.loc[matching_indices, ["integer", "integer.1", "integer.2"]].mean(axis=1).values
            df_all.loc[i+1, 'I-low VT- Tran'] = cb_main.loc[matching_indices, 'I-low VT- Tran'].values
            df_all.loc[i+1, 'I-high VT- Tran'] = cb_main.loc[matching_indices, 'I-high VT- Tran'].values
            df_all.loc[i+1, 'I-amongst-feeders'] = cb_main.loc[matching_indices, 'I-amongst-feeders'].values
            print("heyyy", df_all.loc[i+1])

                    # Thêm hàng vào danh sách
            result_list.append(df_all.loc[[i+1]])  
    
    # Nối tất cả các hàng lại thành một DataFrame mới
    df_result = pd.concat(result_list, ignore_index=True)
    df_result = df_result.drop(columns=["integer"])
    df_result = df_result.drop(columns=["integer.1"])
    df_result = df_result.drop(columns=["integer.2"])
    
    df_result["Label"] = "Replay"
    df_result["Length"] = df["Length"]
    print(df_result)

    
    new_path = path_files_trace.replace("_trace.csv", adding)
    df_result.to_csv(new_path, index=False)
    return df_result
```
