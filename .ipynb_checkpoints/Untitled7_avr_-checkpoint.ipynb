```python
def generate_normal_data_to_csv():    
    paths = glob.glob("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/**/*_trace.csv", recursive=True)
    list_paths =[]
    for p in paths:
        list_paths.append(p)
        
    df10 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED10.csv")
    df13 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/TIED13.csv")
    df30 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED30.csv")
    df31 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED31.csv")
    df32 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED32.csv")
    df33 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED33.csv")
    df41 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED41.csv")
    df42 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED42.csv")
    df43 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED43.csv")

    for d in [df31, df32, df33, df41, df42, df43]:
        d.reset_index(drop=True, inplace=True)
    for path in list_paths:
        df_trace=pd.read_csv(path)
        physical_path = path.replace("_trace", "")
        df = pd.read_csv(physical_path)
        # tính heartbeat (Δt giữa 2 gói liên tiếp cùng goID)
        df_trace["heartbeat"] = df_trace.groupby("goID")["Time"].diff()
        # --- 2. Tính sự thay đổi của stNum ---
        df_trace["dif_stNum"] = df_trace.groupby("goID")["stNum"].diff()
        # --- 3. Tính sự thay đổi của sqNum ---
        df_trace["dif_sqNum"] = df_trace.groupby("goID")["sqNum"].diff()
        # gán hàng loạt (rất nhanh, ít RAM)}
        df_trace = df_trace.drop(columns = ['No.', 'Time', 'integer','integer.1','integer.2'])
        if "UFIED_trace" not in path:
            df_trace['Circuit breaker Open/Close Status'] = np.repeat(df['Circuit breaker Open/Close Status'].values, 3)
            df_trace["I-high VT"] = np.repeat(df10[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]].mean(axis=1).values, 3)
            df_trace["I-high VT- Tran"] = np.repeat(df13[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]].mean(axis=1).values, 3)
            df_trace["I-low VT- Tran"] = np.repeat(df30[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]].mean(axis=1).values, 3)
            df_trace["I-amongst-feeders"] = np.repeat(((df31[" Current Line 'L1'"] + df32[" Current Line 'L1'"] + df33[" Current Line 'L1'"] + df41[" Current Line 'L1'"] + df42[" Current Line 'L1'"] + df43[" Current Line 'L1'"]) / 6).values, 3)
        else:
            df_trace['Circuit breaker Open/Close Status'] = 0
            df_trace["I-high VT"] = np.repeat(df10[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]].mean(axis=1).values, 2)
            df_trace["I-high VT- Tran"] = np.repeat(df13[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]].mean(axis=1).values, 2)
            df_trace["I-low VT- Tran"] = np.repeat(df30[[" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]].mean(axis=1).values, 2)
            df_trace["I-amongst-feeders"] = np.repeat(((df31[" Current Line 'L1'"] + df32[" Current Line 'L1'"] + df33[" Current Line 'L1'"] + df41[" Current Line 'L1'"] + df42[" Current Line 'L1'"] + df43[" Current Line 'L1'"]) / 6).values, 2)
        df_trace = df_trace.fillna(df_trace.mean(numeric_only=True))
        df_trace["Label"] = "Normal"
        combine_path = path.replace("_trace", "_combine")
        df_trace.to_csv(combine_path, index=False)
```

```python
#DOS:AS1
import numpy as np
import pandas as pd
import nest_asyncio
# nest_asyncio.apply()
import csv
import os
import glob
```

```python
path = "IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Attack/Denial of Service (DoS)/AS1_DOS_10.csv"
df_trace = pd.read_csv(path)
# tính heartbeat (Δt giữa 2 gói liên tiếp cùng goID)
df_trace["heartbeat"] = df_trace.groupby("goID")["Time"].diff()
# --- 2. Tính sự thay đổi của stNum ---
df_trace["dif_stNum"] = df_trace.groupby("goID")["stNum"].diff()
# --- 3. Tính sự thay đổi của sqNum ---
df_trace["dif_sqNum"] = df_trace.groupby("goID")["sqNum"].diff()
# gán hàng loạt (rất nhanh, ít RAM)}
df_trace = df_trace.drop(columns = ['Time', 'integer','integer.1','integer.2'])
```

```python
# Hoán đổi hàng 5 và 6 (index = 5, 6)
df_trace.iloc[[530, 531]] = df_trace.iloc[[531, 530]].values
df_trace.iloc[[2675, 2676]] = df_trace.iloc[[2676, 2675]].values
df_trace.iloc[[3794, 3795]] = df_trace.iloc[[3795, 3794]].values

sq = df_trace["sqNum"]
# (1) Gán ID cho từng chuỗi liên tiếp có cùng giá trị
grp = (sq != sq.shift()).cumsum()
# (2) Tính độ dài của mỗi chuỗi
runlen = sq.groupby(grp).transform("size")
# (3) Chọn những chuỗi có độ dài đúng bằng 3
mask_run3 = runlen == 3
# (4) Lấy các chỉ số (index) tương ứng
indices = df_trace.index[mask_run3].tolist()
# (5) Nếu muốn loại bỏ index 3 (tùy mục đích của bạn)
if 3 in indices:
    indices.remove(3)
print(indices)
```

```python
# (7) Lấy ra các giá trị sqNum duy nhất của những chuỗi dài 3
unique_sq_run3 = df_trace.loc[mask_run3, "sqNum"].unique()
print("Unique sqNum (chỉ xuất hiện trong chuỗi dài 3):", unique_sq_run3)
```

```python
# Giả sử đã có 'indices' (đã được sắp xếp)
indices = sorted(indices)

# Tạo nhóm liên tiếp
groups = []
group = [indices[0]]
for i in range(1, len(indices)):
    if indices[i] - indices[i - 1] == 1:
        group.append(indices[i])
    else:
        groups.append(group)
        group = [indices[i]]
groups.append(group)

# Tạo các khoảng trống giữa các nhóm (bao gồm đầu và cuối)
gap_lists = []

# (1) Khoảng trước nhóm đầu tiên
indices_min = 0              # hoặc df_trace.index.min() nếu bạn dùng DataFrame thật
if groups[0][0] > indices_min:
    gap_lists.append(list(range(indices_min, groups[0][0])))

# (2) Khoảng giữa các nhóm
for i in range(len(groups) - 1):
    end_current = groups[i][-1]
    start_next = groups[i + 1][0]
    if start_next - end_current > 1:
        gap_lists.append(list(range(end_current + 1, start_next)))

# (3) Khoảng sau nhóm cuối cùng
indices_max = 5030 - 1       # hoặc df_trace.index.max()
if groups[-1][-1] < indices_max:
    gap_lists.append(list(range(groups[-1][-1] + 1, indices_max + 1)))

print("Các nhóm liên tiếp:", groups)
print("Số nhóm liên tiếp:", len(groups))

len_list = [len(sublist) for sublist in gap_lists]
print("Độ dài từng khoảng:", len_list)
print("Tổng phần tử còn lại:", sum(len_list))
```

```python
df10 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED10.csv")
df13 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/TIED13.csv")
df30 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED30.csv")
df31 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED31.csv")
df32 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED32.csv")
df33 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED33.csv")
df41 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED41.csv")
df42 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED42.csv")
df43 = pd.read_csv("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED43.csv")
physical_path = "IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/Normal/No_Variable_Loading/csv/LIED10.csv"
df = pd.read_csv(physical_path)
```

```python
cols = [" Current Line 'L1'", " Current Line 'L2'", " Current Line 'L3'"]
df_DOS_10_parts = []

for j, k in zip(gap_lists, unique_sq_run3):

    # Lấy bản sao một dòng từ từng df
    df_copy  = df.iloc[[k]]
    df10_copy = df10.iloc[[k]]
    df13_copy = df13.iloc[[k]]
    df30_copy = df30.iloc[[k]]
    df31_copy = df31.iloc[[k]]
    df32_copy = df32.iloc[[k]]
    df33_copy = df33.iloc[[k]]
    df41_copy = df41.iloc[[k]]
    df42_copy = df42.iloc[[k]]
    df43_copy = df43.iloc[[k]]

    # Bản sao dữ liệu trace
    df_trace_copy = df_trace.iloc[j].copy()
    # print(len(df_trace_copy))
    
    # Tính các giá trị scalar
    v_cb = df_copy['Circuit breaker Open/Close Status'].iloc[0]
    v_I_high = df10_copy[cols].mean(axis=1).iloc[0]
    v_I_high_tran = df13_copy[cols].mean(axis=1).iloc[0]
    v_I_low_tran = df30_copy[cols].mean(axis=1).iloc[0]
    v_among = (
        df31_copy[" Current Line 'L1'"].iloc[0] +
        df32_copy[" Current Line 'L1'"].iloc[0] +
        df33_copy[" Current Line 'L1'"].iloc[0] +
        df41_copy[" Current Line 'L1'"].iloc[0] +
        df42_copy[" Current Line 'L1'"].iloc[0] +
        df43_copy[" Current Line 'L1'"].iloc[0]
    ) / 6.0

    # Gán giá trị (broadcast tự động, không cần np.repeat)
    df_trace_copy.loc[:, 'Circuit breaker Open/Close Status'] = v_cb
    df_trace_copy.loc[:, "I-high VT"] = v_I_high
    df_trace_copy.loc[:, "I-high VT- Tran"] = v_I_high_tran
    df_trace_copy.loc[:, "I-low VT- Tran"] = v_I_low_tran
    df_trace_copy.loc[:, "I-amongst-feeders"] = v_among

    # Gộp từng phần
    if len(j) == 528:
        df_DOS_10 = df_trace_copy
    else:
        df_DOS_10 = pd.concat([df_DOS_10, df_trace_copy], axis=0)
df_DOS_10['Label'] = 'DoS'
print("len(df_DOS_10)", len(df_DOS_10))
combine_path = path.replace(".csv", "_combine.csv")
df_DOS_10 = df_DOS_10.drop(columns = ['No.'])
df_DOS_10.to_csv(combine_path, index=False)
```

```python
# Giả sử đã có 'indices' (đã được sắp xếp)
indices = sorted(indices)

# Tạo nhóm liên tiếp
groups = []
group = [indices[0]]
for i in range(1, len(indices)):
    if indices[i] - indices[i - 1] == 1:
        group.append(indices[i])
    else:
        groups.append(group)
        group = [indices[i]]
groups.append(group)

# Tạo các khoảng trống giữa các nhóm (bao gồm đầu và cuối)
gap_lists = []

# (1) Khoảng trước nhóm đầu tiên
indices_min = 0              # hoặc df_trace.index.min() nếu bạn dùng DataFrame thật
if groups[0][0] > indices_min:
    gap_lists.append(list(range(indices_min, groups[0][0])))

# (2) Khoảng giữa các nhóm
for i in range(len(groups) - 1):
    end_current = groups[i][-1]
    start_next = groups[i + 1][0]
    if start_next - end_current > 1:
        gap_lists.append(list(range(end_current + 1, start_next)))

# (3) Khoảng sau nhóm cuối cùng
indices_max = 5039 - 1       # hoặc df_trace.index.max()
if groups[-1][-1] < indices_max:
    gap_lists.append(list(range(groups[-1][-1] + 1, indices_max + 1)))

print("Các nhóm liên tiếp:", groups)
print("Số nhóm liên tiếp:", len(groups))

len_list = [len(sublist) for sublist in gap_lists]
print("Độ dài từng khoảng:", len_list)
print("Tổng phần tử còn lại:", sum(len_list))
```

```python
import pandas as pd
import glob

# Tìm tất cả các file _combine.csv trong thư mục và các thư mục con
paths = glob.glob("IEC61850SecurityDataset-master/IEC61850SecurityDataset-master/**/*_combine.csv", recursive=True)

# Đọc từng file vào DataFrame
df_list = []
for path in paths:
    try:
        df = pd.read_csv(path)
        if "Label" not in df.columns:
            print("⚠️ DataFrame bị thiếu cột 'Label'")
        else:
            print("✅ DataFrame có đầy đủ cột 'Label'")
        df_list.append(df)
        # print(f"Đã đọc xong: {path} ({len(df)} hàng)")
    except Exception as e:
        print(f"Lỗi khi đọc {path}: {e}")

# Nối dọc tất cả các DataFrame
df_all = pd.concat(df_list, axis=0, ignore_index=True)
print(f"\n✅ Tổng cộng {len(df_all)} hàng sau khi nối {len(df_list)} file.")
```

```python
df_all.to_csv("Data.csv", index=False)
```
