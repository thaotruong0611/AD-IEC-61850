```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
import numpy as np
from sklearn.preprocessing import StandardScaler, OrdinalEncoder, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
```

```python
data = pd.read_csv("Data_full.csv")
data.loc[data["goID"].str.startswith("UFIED"), "Circuit breaker Open/Close Status"] = np.nan
x = data.drop(columns = ["Label",  "stNum", "sqNum"])
y = data[["Label"]]
```

```python
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
y_numeric = label_encoder.fit_transform(y)
print(label_encoder.classes_)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
   x, y_numeric, test_size = 0.3, random_state = 42 ,shuffle=True, stratify = y_numeric)
```

```python
num_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("num_scaler", StandardScaler())
])
```

```python
nom_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    # ("HashingEncoder", HashingEncoder(n_components=2))
    ("encoder", OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])
```

```python
# stNum = X_train["stNum"].unique()
Dif_stNum = X_train["Dif_stNum"].unique()
Dif_sqNum = X_train["Dif_sqNum"].unique()
Circuit_breaker_Open_Close_Status_level = [np.sort(X_train["Circuit breaker Open/Close Status"].dropna().unique()).tolist()]
```

```python
ord_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OrdinalEncoder(categories =Circuit_breaker_Open_Close_Status_level, handle_unknown='use_encoded_value', unknown_value=-1))
])
```

```python
preprocessor = ColumnTransformer([
    ("num_feature", num_transformer, ["Length", "Heartbeat", "I-high VT", "I-low VT- Tran", "I-high VT- Tran", "I-amongst-feeders"]),
    ("ord_feature", ord_transformer, ["Circuit breaker Open/Close Status"]),
    ("nom_feature", nom_transformer, ["Source", "Dif_stNum", "Dif_sqNum",])
])
```

```python
from sklearn.svm import SVC

from sklearn.model_selection import GridSearchCV
```

```python
X_train
```

```python
reg = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("regressor", SVC() )
])
```

```python
# params = {
#     "regressor__C":[150],
#     "regressor__kernel":["linear"],
#     "regressor__gamma":["scale"],}

# model = GridSearchCV(
#     estimator=reg,
#     param_grid=params,
#     cv=3
# )
```

```python
params = {
    "regressor__C":[10, 60, 100],
    "regressor__kernel":["linear", "poly", "rbf", "sigmoid", "precomputed"],
    "regressor__gamma":["scale", "auto"],}

model = GridSearchCV(
    estimator=reg,
    param_grid=params,
    cv=3
)
```

```python
model.fit(X_train, y_train)
```
